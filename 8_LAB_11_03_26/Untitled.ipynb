# A* Search for a Puzzle Solver

```python
import heapq
import copy
```

```python
start_state = [
    [1, 2, 3],
    [4, 0, 6],
    [7, 5, 8]
]

goal_state = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]
```

```python
def find_blank(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j
```

```python
def state_to_tuple(state):
    return tuple(tuple(row) for row in state)
```

```python
def h1(state, goal):
    misplaced = 0
    for i in range(3):
        for j in range(3):
            if state[i][j] != 0 and state[i][j] != goal[i][j]:
                misplaced += 1
    return misplaced
```

```python
def h2(state, goal):
    distance = 0

    goal_pos = {}
    for i in range(3):
        for j in range(3):
            goal_pos[goal[i][j]] = (i, j)

    for i in range(3):
        for j in range(3):
            tile = state[i][j]
            if tile != 0:
                goal_i, goal_j = goal_pos[tile]
                distance += abs(i - goal_i) + abs(j - goal_j)

    return distance
```

```python
def get_neighbors(state):
    neighbors = []

    r, c = find_blank(state)

    moves = [(0,1), (1,0), (0,-1), (-1,0)]

    for dr, dc in moves:
        nr, nc = r + dr, c + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_state = copy.deepcopy(state)
            new_state[r][c], new_state[nr][nc] = new_state[nr][nc], new_state[r][c]
            neighbors.append(new_state)

    return neighbors
```

```python
def a_star(start, goal, heuristic):

    pq = []
    heapq.heappush(pq, (heuristic(start, goal), 0, start, [start]))

    visited = set()
    nodes_explored = 0

    while pq:

        f, g, state, path = heapq.heappop(pq)
        nodes_explored += 1

        if state == goal:
            return path, nodes_explored

        visited.add(state_to_tuple(state))

        for neighbor in get_neighbors(state):

            if state_to_tuple(neighbor) not in visited:

                new_g = g + 1
                new_f = new_g + heuristic(neighbor, goal)

                heapq.heappush(
                    pq,
                    (new_f, new_g, neighbor, path + [neighbor])
                )

    return None, nodes_explored
```

```python
path_h1, nodes_h1 = a_star(start_state, goal_state, h1)

print("H1 (Misplaced Tiles)")
print("Nodes Explored:", nodes_h1)
print("Solution Depth:", len(path_h1)-1)
```

```python
path_h2, nodes_h2 = a_star(start_state, goal_state, h2)

print("\nH2 (Manhattan Distance)")
print("Nodes Explored:", nodes_h2)
print("Solution Depth:", len(path_h2)-1)
```

```python
def print_path(path):
    for step in path:
        for row in step:
            print(row)
        print()

print("\nSolution Steps (H2):\n")
print_path(path_h2)
```

```python
print("Performance Comparison\n")

print("H1 Nodes Explored:", nodes_h1)
print("H2 Nodes Explored:", nodes_h2)

print("\nSolution Depth H1:", len(path_h1)-1)
print("Solution Depth H2:", len(path_h2)-1)
```
