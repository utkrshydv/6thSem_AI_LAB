I have attached a file. Please do the following task reading the file using python.
i. arrange the file in the following format
   News Body Content (the large text) \t News Body Headline \t Label 
  and put those in a csv file
ii. Make the count of words contained in News Body Content and News Body
Headline and put the calculated words count in separate columns in
another csv
iii. Make the count of tokens (you can use any tokenizer, like NLTK etc) contained in
News Body Content and News Body Headline and put the calculated token count in separate columns in another csv.

```python
import pandas as pd

filename = 'somefile.txt'

with open(filename, 'r', encoding='utf-8') as f:
    lines = [line.strip() for line in f if line.strip()] 

data = []
for i in range(0, len(lines), 3):

    if i + 2 < len(lines):
        body = lines[i]
        headline = lines[i+1]
        label = lines[i+2]
        data.append([body, headline, label])

df = pd.DataFrame(data, columns=['News Body Content', 'News Body Headline', 'Label'])

df.to_csv('formatted_data.csv', index=False)

print("Part I Complete: Data saved to 'formatted_data.csv'")
df.head()
```

```python
df_words = df.copy()

def count_words(text):
    if pd.isna(text): return 0
    return len(str(text).split())

df_words['Body_Word_Count'] = df_words['News Body Content'].apply(count_words)
df_words['Headline_Word_Count'] = df_words['News Body Headline'].apply(count_words)

output_df_words = df_words[['News Body Content', 'News Body Headline', 'Body_Word_Count', 'Headline_Word_Count']]
output_df_words.to_csv('word_counts.csv', index=False)

print("Part II Complete: Data saved to 'word_counts.csv'")
output_df_words.head()
```

```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab') 

df_tokens = df.copy()

def count_tokens(text):
    if pd.isna(text): return 0
    return len(word_tokenize(str(text)))

df_tokens['Body_Token_Count'] = df_tokens['News Body Content'].apply(count_tokens)
df_tokens['Headline_Token_Count'] = df_tokens['News Body Headline'].apply(count_tokens)

output_df_tokens = df_tokens[['News Body Content', 'News Body Headline', 'Body_Token_Count', 'Headline_Token_Count']]
output_df_tokens.to_csv('token_counts.csv', index=False)

print("Part III Complete: Data saved to 'token_counts.csv'")
output_df_tokens.head()
```
