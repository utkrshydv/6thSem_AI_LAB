```python
import heapq
```

```python
def heuristic(node, goal):
    return abs(node[0] - goal[0]) + abs(node[1] - goal[1])
```

```python
def get_neighbors(node, size, blocked):
    x, y = node
    neighbors = []
    
    moves = [(1,0), (-1,0), (0,1), (0,-1)]
    
    for dx, dy in moves:
        nx, ny = x + dx, y + dy
        
        if 0 <= nx < size and 0 <= ny < size:
            if (nx, ny) not in blocked:
                neighbors.append((nx, ny))
                
    return neighbors
```

```python
def best_first_search(start, goal, size, blocked):
    
    visited = set()
    priority_queue = []
    parent = {}
    
    heapq.heappush(priority_queue, (heuristic(start, goal), start))
    
    while priority_queue:
        
        h, current = heapq.heappop(priority_queue)
        
        if current in visited:
            continue
        
        print("Visiting:", current, "Heuristic:", h)
        visited.add(current)
        
        if current == goal:
            print("\nTreasure Found!")
            return parent
        
        for neighbor in get_neighbors(current, size, blocked):
            if neighbor not in visited:
                heapq.heappush(priority_queue, (heuristic(neighbor, goal), neighbor))
                parent[neighbor] = current
    
    print("\nNo Path Found!")
    return parent
```

```python
def reconstruct_path(parent, start, goal):
    
    if goal not in parent:
        print("Cannot reconstruct path.")
        return []
    
    path = []
    current = goal
    
    while current != start:
        path.append(current)
        current = parent[current]
    
    path.append(start)
    path.reverse()
    
    return path
```

```python
choice = input("Enter 1 for Default Example OR 2 for Custom Input: ")

if choice == "1":
    
    # Default Example
    size = 4
    start = (0,0)
    goal = (3,3)
    blocked = {(1,0), (1,1), (1,3), (3,1), (3,2)}
    
    print("\nRunning Default Example...\n")

elif choice == "2":
    
    # Custom Input
    size = int(input("Enter grid size (n for n x n): "))
    
    sx, sy = map(int, input("Start X & Start Y: ").split(", "))
    start = (sx, sy)
    
    gx, gy = map(int, input("Goal X & Goal Y: ").split(", "))
    goal = (gx, gy)
    
    b = int(input("How many blocked cells? "))
    blocked = set()
    
    for i in range(b):
        x,y  = map(int, input(f"Blocked cell {i+1} - X & Y: ").split(", "))
        blocked.add((x,y))


else:
    print("Invalid Choice!")
```

```python
parent = best_first_search(start, goal, size, blocked)

path = reconstruct_path(parent, start, goal)

print("\nFinal Path:", path)
```

```python
#visualisation
visual = []
for r in range(size):
    row = []
    for c in range(size):
        if (r,c) in blocked:
            row.append('0')
        else:
            row.append('1')
    visual.append(row)
for (r, c) in path:
    if (r,c) == (sx, sy):
        visual[r][c] = "S"
    elif (r,c) == (gx, gy):
        visual[r][c] = "G"
    else: 
        visual[r][c] = "*"

print("\nBest First Search Visualization: \n")
for row in visual:
    print(" ".join(row))
```
